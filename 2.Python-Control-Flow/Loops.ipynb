## Loops in Python


```python
## for loop

range(1,5)
```

```python
for i in range(1,5):
    print(i)
```

```python
# this is a step where 1 is the start idx and 20 is the stop idx and 2 is the step idx

for i in range(1,20,2):
    print(i)
```

```python
# we can even print in reverse order

for i in range(10,1,-2):
    print(i)
```

```python
# string
str = "Pranith Reddy"

for i in str:
    print(i)
```

```python
# while loop

# while loop continues to exist as long as the condition is True

count = 0
while count<5:
    print(count)
    count=count+1


```

```python
# break statement 

# break exits the loop prematurely
for i in range(1,10):
    if i ==5:
        break
    print(i) # it won't print value of 5 nad after
```

```python
# continue statement

# the continue statement skips the current iteration and continues with the next 

for i in range(5):
    if i%2==0:
        continue
    print(i) # only print the odd number as the even number is skipped
```

```python
# pass statement 

# the pass statement does null operation meaning it does nothing

for i in range(10):
    if i==7:
       # print("the null is for ",i) # this is the null or pass statement execution for number 7
        pass
    print(i)
```

```python
## nested loops

for i in range(3):
    for j in range(2):
        print(f" i:{i}  j:{j} ")
```

```python
# calculate the sum of natural numbers using while and for loop

# using formulae
num = int(input("Enter a number"))
print("Result :",num*(num+1)//2)

#using for loop

sum=0
for i in range(num+1):
    sum = sum+i
print("Sum is :",sum)

#using while loop

result = 0
ans=0
while result<=num:
    ans = ans+result
    result = result +1
print("Ans is :",ans)
```

```python
# if a num is prime or not from 1 to 100

num = int(input("Enter a number"))

for num in range(1,num):
    if num>1:
        for i in range(2,num):
            if num % i == 0:
                break
        else:
            print(num)

        
```
