## Conditional Statements

```python
# if statement

age =20

if age>=18:
    print("Eligible to vote")
```

```python
age>=18
```

```python
# else statement is used or declared to be executed when the if statement is a false condition

age =16

if age>=18:
    print("eligible to vote")
else:
    print("you are a minor")
```

```python
# elif this is a condition which can allow you to check multiple conditions also called as else if

age=15

if age<=13:
    print("you are a minor")
elif age<=16:
    print("you are a teen")
else:
    print("you are a major")
```

```python
# nested Conditional Statements

num = int(input("Enter the number"))

if num>0:
    print("the number is positive")
    if num%2==0:
        print("the number is even")
    else:
        print("the number is odd")
else:
    print("the number is zero or negative")
```

```python
# determine if a year is a leap year using nested conditional statement 

year = int(input("Enter the year: "))

if year % 400 == 0:
    print(f"{year} is a leap year")
elif year % 100 == 0:
    print(f"{year} is not a leap year")
elif year % 4 == 0:
    print(f"{year} is a leap year")
else:
    print(f"{year} is not a leap year")   
```

```python
year = int(input("Enter the year: "))

if year%4==0:
    if year%100==0:
        if year%400==0:
            print(f"{year} is leap year")
        else:
            print(f"{year} is not a leap year")
    else :
        print(f"{year} is a leap year")
else :
    print(f"{year} is not a leap year")
```

```python
# Simple calculator program

num1 = float(input("Enter a number"))
num2 = float(input("Enter another number"))

operation = input("Enter any operand between (+ ,- , * , /)")

if operation == '+':
    print(f"addition is {num1 + num2}")
elif operation == '-':
    print(f"subtraction is {num1-num2}")
elif operation == '*':
    print(f"multiplication is {num1 * num2}")
elif operation =='/':
    if num2 !=0:
        print(f"division is {num1/num2}")
    else:
        print("error cannot divide by zero")
else :
    print("not a valid number or operation")
```
