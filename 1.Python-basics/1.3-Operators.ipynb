##  Operators 

### Arithmetic Operators

```python
## arthmetic operators

a=10
b=5

add = a+b
sub = a-b
div = a/b
mul = a*b
floor_div = a//b   # basically removes the decimal part of the number or rounding of to the lowest value
exp = a**b
modulus = a%b

print(add)
print(sub)
print(div)
print(mul)
print(floor_div)
print(exp)
print(modulus)
```

### Logical Operators

```python
# and operator

## in and operator the both the variables should be true to get a true value if any of the value is false the result is false

a=True
b=False

result = a and b
print(result)
```

```python
a=True
b=True

result = a and b
print(result)
```

```python
# OR operator

# if any one operator is true then the result is true
a=True
b=False

result = a or b
print(result)
```

```python
a=False
b=False

result = a and b
print(result)
```

```python
# Not operator

a=True
not a
```
