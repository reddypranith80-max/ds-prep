```python
## Understanding Variable types

# Python is Dynamically typed , the type of variable is declared at runtime

age=22 #int
height=2.2 #float
name="Pranith" #str
is_Student=True #bool
```

```python
# type checking and conversion

age = 25
print(type(age))

age_str = str(age)
print(age_str)
print(type(age_str))
```

```python
age='25'
print(type(age))

print(int(age)) # converting 25 which is string to int
print(type(int(age)))
```

```python
name ="Pranith"

print(int(name)) # the conversion of string pranith cannot be converted to integer
```

```python
height = 2.2
print(type(height))

int(height) # we can convert the float to int by rounding off

float(int(height)) # before it was 2.2 now it is 2.0 loss in conversion from float->int and int->float
```

```python
## dynamic typing
# here in python we can change the variable and it's type as the program executes

var = 2
print(var,type(var))

var = "Pranith"
print(var,type(var))

var = 3.14
print(var,type(var))
```

```python
# input

var = input("Enter the number")
print(var,type(var))

var = int(input("enter the num"))
print(var,type(var))
```
